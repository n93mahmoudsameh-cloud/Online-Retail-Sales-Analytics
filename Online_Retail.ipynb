Step 4 : Data Cleaning

```python
# Remove Duplicate Values
Online_Retail.drop_duplicates(inplace=True)
```

```python
print('Duplicate_Value : ' ,Online_Retail.duplicated().sum())
```

```python
# Handling Missing Values
Online_Retail.dropna(inplace=True)
```

```python
print('Null_Values Per Columns : \n' ,Online_Retail.isnull().sum())
print('Total Null_Values : ' ,Online_Retail.isnull().sum().sum())
```

```python
# Change Data Type
Online_Retail['InvoiceDate'] = pd.to_datetime(Online_Retail['InvoiceDate'])
```

```python

Online_Retail['Country']=Online_Retail['Country'].replace({
    'EIR':'Ireland',
    'RSA':'South Africa',
    'USA':'United States',
    'European Community':'United Kingdom',
    'Unspecified':'United Kingdom'})
Online_Retail['Country']=Online_Retail['Country'].astype('category')
```

```python
Online_Retail.info()
```

Step 5 : Feature Engineering

```python
# Export Total_Sales

Online_Retail['Total Sales']=(
Online_Retail['Price']*Online_Retail['Quantity'])
Online_Retail['Total Sales']=round(Online_Retail['Total Sales'],2)
```

```python
# Export Year, Month, Quarter & Day

Online_Retail['Year']=Online_Retail['InvoiceDate'].dt.year
Online_Retail['Month']=Online_Retail['InvoiceDate'].dt.month
Online_Retail['Month_Name']=Online_Retail['InvoiceDate'].dt.month_name()
Online_Retail['Quarter']=Online_Retail['InvoiceDate'].dt.quarter
Online_Retail['Day']=Online_Retail['InvoiceDate'].dt.day
Online_Retail['Day_Name']=Online_Retail['InvoiceDate'].dt.day_name()
```
